```python
class Node:
  def __init__(self,data):
    self.data = data
    self.right = None
    self.left = None

class Tree:
  def __init__(self):
    self.root = None

  def insert(self,val):
    if self.root is None:
      self.root = Node(val)
      return

    temp = self.root
    while True:

      if temp.data == val:
        return
      elif val > temp.data and temp.right is None:
        temp.right = Node(val)
        return

      elif val < temp.data and temp.left is None:
        temp.left = Node(val)
        return

      elif val > temp.data:
        temp = temp.right
      else:
        temp = temp.left

  def traversal(self,root):
    if root is None:
      return

    self.traversal(root.left)
    print(root.data)
    self.traversal(root.right)

s = Tree()
arr=[1,2,4,5,6,7,3,8]
for i in arr:
  s.insert(i)
s.traversal(s.root)
```

```python
class Solution:
def sortedListToBST(self, head: Optional[ListNode]) -> Optional[TreeNode]:

    if head is None:
        return None

    if head.next is None:
        return TreeNode(head.val)

    slow = head
    fast = head
    prev = None

    while fast and fast.next:
        prev = slow
        slow = slow.next
        fast = fast.next.next

    prev.next = None

    root = TreeNode(slow.val)

    root.left = self.sortedListToBST(head)
    root.right = self.sortedListToBST(slow.next)
```

```python
nums = [5, 3, 8, 4, 2]

def select(nums):
  n = len(nums)
  for i in range(n):
    min = i
    for j in range(i+1,n):
      if nums[j] < nums[min]:
       min = j
    if min != i:
      nums[i],nums[min] = nums[min],nums[i]
  return nums
print(select(nums))
```

```python
nums = [5, 3, 8, 4, 2]

def bubble(nums):
  n = len(nums)

  for i in range(n-1):
    for j in range(n-1-i):
      if nums[j] > nums[j+1]:
        nums[j],nums[j+1] = nums[j+1],nums[j]
  return nums
print(bubble(nums))
```

```python
def findMinimumSum(power):
    res = 0
    pfinal = power[0]
    pinc = 0

    for i in range(1, len(power)):
        curr = power[i]
        inc = max(0, pfinal - curr)
        res += max(0, inc - pinc)
        pinc = inc
        pfinal = max(pfinal, curr + inc)

    return res
arr = [3,5,2,3]
print(findMinimumSum(arr))
```

```python
from itertools import combinations
s2 = "ace"
def get_all_subsequences(string):
    subsequences = []

    for r in range(len(string) + 1):
        for combo in combinations(string, r):
            subsequences.append("".join(combo))

    return subsequences


res = list(get_all_subsequences("abcde"))
if s2 in res:
  print(len(res[res.index(s2)]))
```

```python
def pasha_maximizes():
    # Read inputs
    s, k = input().split()
    digits = list(s)
    k = int(k)
    n = len(digits)

    for i in range(n):
        if k <= 0:
            break

        max_idx = i
        # Check elements we can reach with remaining k swaps
        for j in range(i + 1, min(n, i + k + 1)):
            if digits[j] > digits[max_idx]:
                max_idx = j

        # Bring the largest found digit to the current position i
        for j in range(max_idx, i, -1):
            digits[j], digits[j - 1] = digits[j - 1], digits[j]
            k -= 1

    # print("".join(digits))

if __name__ == "__main__":
    pasha_maximizes()
```
