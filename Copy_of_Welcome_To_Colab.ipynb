```
#1723
class Solution:
    def minimumTimeRequired(self, jobs, k):

        jobs.sort(reverse=True)

        low = max(jobs)
        high = sum(jobs)

        def can_finish(limit):

            workers = [0] * k

            def dfs(index):

                if index == len(jobs):
                    return True

                for i in range(k):

                    if workers[i] + jobs[index] <= limit:

                        workers[i] += jobs[index]


                        if dfs(index + 1):
                            return True

                        workers[i] -= jobs[index]

                    if workers[i] == 0:
                        break

                return False

            return dfs(0)

        while low < high:

            mid = (low + high) // 2

            if can_finish(mid):
                high = mid
            else:
                low = mid + 1

        return low
```

```
#1494
from functools import lru_cache
from itertools import combinations

class Solution:
    def minNumberOfSemesters(self, n, relations, k):

        pre = [0] * n

        for u, v in relations:
            pre[v - 1] |= 1 << (u - 1)

        FULL = (1 << n) - 1

        @lru_cache(None)
        def dfs(mask):

            if mask == FULL:
                return 0

            available = []

            for course in range(n):

                if mask & (1 << course):
                    continue

                if (pre[course] & mask) == pre[course]:
                    available.append(course)

            if len(available) <= k:
                nxt = mask
                for c in available:
                    nxt |= 1 << c
                return 1 + dfs(nxt)

            ans = float('inf')

            for subset in combinations(available, k):

                nxt = mask

                for c in subset:
                    nxt |= 1 << c

                ans = min(ans, 1 + dfs(nxt))

            return ans

        return dfs(0)
```

```
#1434 ways to wear different hats
from functools import lru_cache
from collections import defaultdict

class Solution:
    def numberWays(self, hats):
        MOD = 10**9 + 7
        n = len(hats)

        hat_to_people = defaultdict(list)

        for person in range(n):
            for hat in hats[person]:
                hat_to_people[hat].append(person)

        target = (1 << n) - 1

        @lru_cache(None)
        def dfs(hat, mask):

            if mask == target:
                return 1

            if hat > 40:
                return 0

            ans = dfs(hat + 1, mask)

            for person in hat_to_people[hat]:

                if mask & (1 << person):
                    continue

                ans += dfs(hat + 1,
                           mask | (1 << person))

            return ans % MOD

        return dfs(1, 0)
```

```
#1125 sufficient team
class Solution:
    def smallestSufficientTeam(self, req_skills, people):

        skill_id = {}

        for i, skill in enumerate(req_skills):
            skill_id[skill] = i

        person_masks = []

        for person in people:
            mask = 0
            for skill in person:
                if skill in skill_id:
                    mask |= 1 << skill_id[skill]
            person_masks.append(mask)

        dp = {0: []}

        for i, pmask in enumerate(person_masks):

            current = list(dp.items())

            for mask, team in current:

                newmask = mask | pmask

                if newmask not in dp or len(dp[newmask]) > len(team) + 1:
                    dp[newmask] = team + [i]

        full = (1 << len(req_skills)) - 1

        return dp[full]
```

```
# 864   shortest path for getting all keys
from collections import deque

class Solution:
    def shortestPathAllKeys(self, grid):
        m, n = len(grid), len(grid[0])

        start = None
        key_count = 0


        for i in range(m):
            for j in range(n):
                if grid[i][j] == '@':
                    start = (i, j)
                elif 'a' <= grid[i][j] <= 'f':
                    key_count += 1

        target = (1 << key_count) - 1

        queue = deque()
        queue.append((start[0], start[1], 0, 0))

        visited = set()
        visited.add((start[0], start[1], 0))

        directions = [(1,0),(-1,0),(0,1),(0,-1)]

        while queue:
            r, c, mask, dist = queue.popleft()

            if mask == target:
                return dist

            for dr, dc in directions:
                nr, nc = r + dr, c + dc

                if not (0 <= nr < m and 0 <= nc < n):
                    continue

                cell = grid[nr][nc]

                if cell == '#':
                    continue

                new_mask = mask


                if 'a' <= cell <= 'f':
                    new_mask |= 1 << (ord(cell) - ord('a'))


                if 'A' <= cell <= 'F':
                    if not (new_mask & (1 << (ord(cell) - ord('A')))):
                        continue

                if (nr, nc, new_mask) not in visited:
                    visited.add((nr, nc, new_mask))
                    queue.append((nr, nc, new_mask, dist + 1))

        return -1
```

```
# 847 # shortest path visiting all Nodes
from collections import deque

class Solution:
    def shortestPathLength(self, graph):
        n = len(graph)

        target = (1 << n) - 1

        queue = deque()
        visited = set()


        for i in range(n):
            mask = 1 << i
            queue.append((i, mask, 0))
            visited.add((i, mask))

        while queue:
            node, mask, dist = queue.popleft()

            if mask == target:
                return dist

            for nei in graph[node]:
                new_mask = mask | (1 << nei)

                if (nei, new_mask) not in visited:
                    visited.add((nei, new_mask))
                    queue.append((nei, new_mask, dist + 1))
```

```
#------------------- unique path 980------------------#
class Solution:
    def uniquePathsIII(self, grid):
        rows = len(grid)
        cols = len(grid[0])

        total = 0


        for i in range(rows):
            for j in range(cols):
                if grid[i][j] != -1:
                    total += 1

                if grid[i][j] == 1:
                    sr, sc = i, j

        ans = 0

        def dfs(r, c, count):
            nonlocal ans

            if (r < 0 or c < 0 or
                r >= rows or c >= cols or
                grid[r][c] == -1):
                return

            # Reached the end
            if grid[r][c] == 2:
                if count == total:
                    ans += 1
                return

            # Mark as visited
            temp = grid[r][c]
            grid[r][c] = -1

            # Explore 4 directions
            dfs(r + 1, c, count + 1)
            dfs(r - 1, c, count + 1)
            dfs(r, c + 1, count + 1)
            dfs(r, c - 1, count + 1)

            # Backtrack
            grid[r][c] = temp

        dfs(sr, sc, 1)

        return ans
```

```
#-------------------->squared 996 ------------------
from math import isqrt

class Solution:
    def numSquarefulPerms(self, nums):
        nums.sort()
        n = len(nums)
        used = [False] * n
        ans = 0

        def isSquare(x):
            r = isqrt(x)
            return r * r == x

        def backtrack(path):
            nonlocal ans

            if len(path) == n:
                ans += 1
                return

            for i in range(n):

                if used[i]:
                    continue

                # Skip duplicate numbers
                if i > 0 and nums[i] == nums[i-1] and not used[i-1]:
                    continue

                # Adjacent sum must be a perfect square
                if path and not isSquare(path[-1] + nums[i]):
                    continue

                used[i] = True
                path.append(nums[i])

                backtrack(path)

                path.pop()
                used[i] = False

        backtrack([])

        return ans
```

```
def subsequence(s):
  result = []
  def backtrack(index,curr):
    if index == len(s):
      result.append(curr)
      return

    backtrack(index+1, curr+s[index])
    backtrack(index+1,curr)

  backtrack(0,"")
  return result

if __name__ == "__main__":
  print(subsequence("abc"))
```

```
# =========== Stack Example ===========#
class Stack:
  def __init__(self):
    self.stack = []

  def push(self,element):
    self.stack.append(element)

  def pop(self):
    if self.isEmpty():
      return "Stack is Empty"
    else:
      return self.stack.pop()

  def peek(self):
    return self.stack[-1]

  def isEmpty(self):
    return len(self.stack) == 0

  def size(self):
    return len(self.stack)

mystack = Stack()

list = [10,20,30,40,50,60,70]

for i in list:
  mystack.push(i)

print(mystack.stack)
print(mystack.peek())
print(mystack.isEmpty())
print(mystack.size())
```

```
# =========== Queue Example ===========#
class Queue:

  def __init__(self):
    self.queue = []

  def Enqueue(self,element):
    self.queue.append(element)

  def Dequeue(self):
    if self.isEmpty():
      return "Queue is Empty"
    else:
      return self.queue.pop()

  def Peek(self):
    return self.queue[0]

  def isEmpty(self):
    return len(self.queue)

  def Size(self):
    return len(self.queue)

my_queue = Queue()

list = [10,20,30,40,50,60,70]
for i in list:
  my_queue.Enqueue(i)

print(my_queue.queue)
print(my_queue.Size())
```

```
#======== Deleting a value in Singly Linked List ========#

class Node:
  def __init__(self,data):
    self.data = data
    self.next = None

def print_items(head):
  current = head

  while current:
    print(current.data,end='->')
    current = current.next
  print("Null")

def delete_items(head,deleteitem):
  if head == deleteitem:
    return head.next

  current = head
  while(current.next and current.next != deleteitem):
    current = current.next

  if current.next is None:
    return head

  current.next = current.next.next
  return head

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)
node5 = Node(50)
node6 = Node(60)

node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5
node5.next = node6

print("List Before Deletion:")
print_items(node1)
node1 = delete_items(node1,node6)
print("List Before Deletion:")
print_items(node1)
```

```
#==== Binary Search =====#

def binarySearch(arr, target):

  low = 0
  high = len(arr)-1

  while low <= high:
    mid = low +(high - low)//2
    if arr[mid] == target:
      return mid
    elif arr[mid] < target:
      low = mid+1
    else:
      high = mid-1
  return -1

mylist = list(map(int,input().split()))
target = 5
result = binarySearch(mylist, target)

if result != -1:
  print(f"Found at:{result}")
else:
  print("Not Found")
```

```
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

def printList(head):
    current = head
    while current:
        print(current.data, end="->")
        current = current.next
    print("Null")

# --- DELETION LOGIC ---

def deleteAtLast(head):
    # Edge Case 1: List is empty
    if head is None:
        return None

    # Edge Case 2: Only one node in the list
    if head.next is None:
        return None

    # Traverse to the second-to-last node
    temp = head
    while temp.next.next is not None:
        temp = temp.next

    # Disconnect the last node
    temp.next = None
    return head

def deleteAtMiddle(head, position):
    # Edge Case 1: List is empty
    if head is None:
        print("List is empty")
        return None

    # Edge Case 2: Deleting the head node (position 0)
    if position == 0:
        return head.next

    temp = head
    # Traverse to the node right BEFORE the one we want to delete
    for i in range(position - 1):
        if temp is None or temp.next is None:
            break
        temp = temp.next

    # If the position is out of bounds
    if temp is None or temp.next is None:
        print("Position out of bounds")
        return head

    # Skip the target node
    temp.next = temp.next.next
    return head

# --- MAIN EXECUTION ---

def main():
    head = None
    tail = None

    print("Enter elements (-1 to stop, space-separated):")
    input_str = input()
    values = [int(s) for s in input_str.split()]

    for val in values:
        if val == -1:
            break

        newNode = Node(val)
        if head is None:
            head = newNode
            tail = newNode
        else:
            tail.next = newNode
            tail = newNode

    print("\nOriginal List:")
    printList(head)

    # Example 1: Delete at Last
    print("\nAfter deleting last node:")
    head = deleteAtLast(head)
    printList(head)

    # Example 2: Delete at Index 1 (Middle element if list has 3+ elements)
    print("\nAfter deleting node at index 1:")
    head = deleteAtMiddle(head, 1)
    printList(head)

if __name__ == "__main__":
    main()
```
